# 1. 학생 성적 데이터 분석 예제

### 목표
- 과목 평균 구하기
- 최고 점수 학생 찾기
- 특정 조건 학생 찾기
- 함수로 분석 기능 분리하기


```python
students = [
    {"name": "민준", "kor": 88, "eng": 92, "math": 90},
    {"name": "서연", "kor": 95, "eng": 89, "math": 98},
    {"name": "지우", "kor": 76, "eng": 85, "math": 80},
    {"name": "하준", "kor": 90, "eng": 91, "math": 84},
    {"name": "유나", "kor": 100, "eng": 94, "math": 96}
]

print(students)
```

### 예제 1-1. 과목 평균 구하기

```python
def subject_average(data, subject):
    total = 0
    for student in data:
        total += student[subject]
    return total / len(data)

print("국어 평균:", subject_average(students, "kor"))
print("영어 평균:", subject_average(students, "eng"))
print("수학 평균:", subject_average(students, "math"))
```

### 예제 1-2. 총점이 가장 높은 학생 찾기

```python
def total_score(student):
    return student["kor"] + student["eng"] + student["math"]

top_student = students[0]
top_total = total_score(top_student)

for student in students:
    if total_score(student) > top_total:
        top_student = student
        top_total = total_score(student)

print("총점 최고 학생:", top_student["name"])
print("총점:", top_total)
```

### 예제 1-3. 수학 90점 이상 학생 찾기

```python
high_math = []

for student in students:
    if student["math"] >= 90:
        high_math.append(student["name"])

print("수학 90점 이상 학생:", high_math)
```

# 2. 편의점 판매 데이터 분석 예제

### 목표
- 전체 매출 구하기
- 상품별 판매량 집계
- 가장 많이 팔린 상품 찾기
- 특정 상품 총매출 계산하기


```python
sales = [
    {"item": "삼각김밥", "price": 1200, "qty": 3},
    {"item": "주스", "price": 1800, "qty": 2},
    {"item": "과자", "price": 1500, "qty": 5},
    {"item": "삼각김밥", "price": 1200, "qty": 2},
    {"item": "우유", "price": 1000, "qty": 4},
    {"item": "주스", "price": 1800, "qty": 1}
]

print(sales)
```

### 예제 2-1. 전체 매출 계산

```python
total_sales = 0

for row in sales:
    total_sales += row["price"] * row["qty"]

print("전체 매출:", total_sales)
```

### 예제 2-2. 상품별 판매량 집계

```python
item_count = {}

for row in sales:
    item = row["item"]
    qty = row["qty"]

    if item in item_count:
        item_count[item] += qty
    else:
        item_count[item] = qty

print("상품별 판매량:", item_count)
```

### 예제 2-3. 가장 많이 팔린 상품 찾기

```python
best_item = ""
best_qty = 0

for item in item_count:
    if item_count[item] > best_qty:
        best_item = item
        best_qty = item_count[item]

print("가장 많이 팔린 상품:", best_item)
print("판매 개수:", best_qty)
```

# 3. 취미 데이터 집합 분석 예제

### 목표
- 중복 제거
- 공통 취미 찾기
- 전체 취미 종류 찾기
- 취미 빈도수 세기


```python
hobby_data = {
    "민준": ["축구", "게임", "영화"],
    "서연": ["독서", "영화", "음악"],
    "지우": ["게임", "축구", "음악"],
    "하준": ["축구", "게임", "독서"],
    "유나": ["영화", "음악", "독서"]
}

print(hobby_data)
```

### 예제 3-1. 전체 취미 종류 구하기

```python
all_hobbies = set()

for student in hobby_data:
    hobbies = hobby_data[student]
    for hobby in hobbies:
        all_hobbies.add(hobby)

print("전체 취미 종류:", all_hobbies)
```

### 예제 3-2. 두 학생의 공통 취미 찾기

```python
student1 = "민준"
student2 = "하준"

common = set(hobby_data[student1]) & set(hobby_data[student2])

print(f"{student1}와 {student2}의 공통 취미:", common)
```

### 예제 3-3. 취미 빈도수 세기

```python
hobby_count = {}

for student in hobby_data:
    for hobby in hobby_data[student]:
        if hobby in hobby_count:
            hobby_count[hobby] += 1
        else:
            hobby_count[hobby] = 1

print("취미 빈도수:", hobby_count)
```

# 4. 날씨 데이터 분석 예제

### 목표
- 평균 기온 구하기
- 최고기온 / 최저기온 찾기
- 조건에 맞는 데이터 추출


```python
weather = [
    {"day": "월", "temp": 22},
    {"day": "화", "temp": 24},
    {"day": "수", "temp": 21},
    {"day": "목", "temp": 26},
    {"day": "금", "temp": 23}
]

print(weather)
```

### 예제 4-1. 평균 기온 구하기

```python
total = 0

for row in weather:
    total += row["temp"]

avg = total / len(weather)
print("평균 기온:", avg)
```

### 예제 4-2. 가장 더운 날 찾기

```python
hottest = weather[0]

for row in weather:
    if row["temp"] > hottest["temp"]:
        hottest = row

print("가장 더운 날:", hottest["day"])
print("기온:", hottest["temp"])
```

# 5. 검색어 빈도 분석 예제

### 목표
- 검색어 빈도수 세기
- 가장 많이 나온 검색어 찾기
- 한 번만 나온 검색어 찾기


```python
keywords = ["AI", "파이썬", "AI", "데이터", "파이썬", "AI", "코딩"]

print(keywords)
```

### 예제 5-1. 검색어 빈도수 세기

```python
count = {}

for word in keywords:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

print("검색어 빈도수:", count)
```

### 예제 5-2. 가장 많이 나온 검색어 찾기

```python
best_word = ""
best_count = 0

for word in count:
    if count[word] > best_count:
        best_word = word
        best_count = count[word]

print("가장 많이 나온 검색어:", best_word)
print("횟수:", best_count)
```
